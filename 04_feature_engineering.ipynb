```python
import pandas as pd
df = pd.read_csv("cleaned_supply_chain_data.csv")
```

```python
df.columns
```

```python
df.select_dtypes(include="number").columns
```

```python
df["Total Cost"] = df["Shipping costs"] + df["Manufacturing costs"]
```

```python
df["Revenue per Product"] = df["Revenue generated"] / df["Number of products sold"]
```

```python
df["Cost per Product"] = df["Total Cost"] / df["Number of products sold"]
```

```python
df["Estimated Profit"] = df["Revenue generated"] - df["Total Cost"]
```

```python
df["Profit Margin"] = (df["Estimated Profit"] / df["Revenue generated"]) * 100
```

```python
df[[
    "Total Cost",
    "Revenue per Product",
    "Cost per Product",
    "Estimated Profit",
    "Profit Margin"
]].head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df["Sales Efficiency"] = (
    df["Number of products sold"] / df["Production volumes"]
)
```

```python
df["Inventory Turnover"] = (
    df["Number of products sold"] / df["Stock levels"]
)
```

```python
df["Order Fulfillment Ratio"] = (
    df["Number of products sold"] / df["Order quantities"]
)
```

```python
df["Shipping Delay"] = df["Shipping times"] - df["Lead times"]
```

```python
df["Estimated Defect Quantity"] = (
    df["Production volumes"] * df["Defect rates"] / 100
)
```

```python
df["Cost Percentage"] = (
    df["Total Cost"] / df["Revenue generated"]
) * 100
```

```python
new_features = [
    "Total Cost",
    "Revenue per Product",
    "Cost per Product",
    "Estimated Profit",
    "Profit Margin",
    "Sales Efficiency",
    "Inventory Turnover",
    "Order Fulfillment Ratio",
    "Shipping Delay",
    "Estimated Defect Quantity",
    "Cost Percentage"
]

df[new_features].head()
```

```python
import numpy as np

df.replace([np.inf, -np.inf], np.nan, inplace=True)

df[new_features].isnull().sum()
```

```python
df.to_csv("feature_engineered_supply_chain_data.csv", index=False)
```
