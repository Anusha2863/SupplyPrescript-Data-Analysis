```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_supply_chain_data.csv")
df.head()
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numerical_columns)
```

```python
df[numerical_columns].describe()
```

```python
categorical_columns = df.select_dtypes(include='object').columns

print("Categorical Columns:")
print(categorical_columns)
```

```python
df[categorical_columns].nunique()
```

```python
for column in categorical_columns:
    print("\n" + column)
    print(df[column].unique())
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Revenue generated'], bins=10)
plt.xlabel('Revenue Generated')
plt.ylabel('Frequency')
plt.title('Distribution of Revenue Generated')
plt.show()
print("Observation: The histogram shows how the revenue values are distributed across the dataset.")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Price'], bins=10)
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Product Price')
plt.show()
```

```python
revenue_by_product = df.groupby('Product type')['Revenue generated'].mean()

plt.figure(figsize=(8, 5))
revenue_by_product.plot(kind='bar')
plt.xlabel('Product Type')
plt.ylabel('Average Revenue')
plt.title('Average Revenue by Product Type')
plt.xticks(rotation=0)
plt.show()
print("Business Insight: This chart helps identify which product type generates the highest average revenue.")
```

```python
sales_by_product = df.groupby('Product type')['Number of products sold'].sum()

plt.figure(figsize=(8, 5))
sales_by_product.plot(kind='bar')
plt.xlabel('Product Type')
plt.ylabel('Number of Products Sold')
plt.title('Products Sold by Product Type')
plt.xticks(rotation=0)
plt.show()
print("Business Insight: Product types with higher sales volume may require better inventory planning.")
```

```python
revenue_by_demographics = df.groupby('Customer demographics')['Revenue generated'].mean()

plt.figure(figsize=(8, 5))
revenue_by_demographics.plot(kind='bar')
plt.xlabel('Customer Demographics')
plt.ylabel('Average Revenue')
plt.title('Average Revenue by Customer Demographics')
plt.xticks(rotation=0)
plt.show()
print("Business Insight: This chart helps identify customer groups that contribute more to revenue.")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Manufacturing costs'], bins=10)
plt.xlabel('Manufacturing Cost')
plt.ylabel('Frequency')
plt.title('Distribution of Manufacturing Costs')
plt.show()
print("Business Insight: The distribution helps identify the typical manufacturing cost range and unusually high or low costs.")
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df['Revenue generated'])
plt.ylabel('Revenue Generated')
plt.title('Boxplot of Revenue Generated')
plt.show()
print("Observation: The boxplot helps identify the spread of revenue and potential outliers.")
```

```python
product_counts = df['Product type'].value_counts()

plt.figure(figsize=(8, 5))
product_counts.plot(kind='bar')
plt.xlabel('Product Type')
plt.ylabel('Count')
plt.title('Number of Records by Product Type')
plt.xticks(rotation=0)
plt.show()
print("Business Insight: This chart shows how the products are distributed across different product types.")
```

```python
transport_counts = df['Transportation modes'].value_counts()

plt.figure(figsize=(8, 5))
transport_counts.plot(kind='bar')
plt.xlabel('Transportation Mode')
plt.ylabel('Count')
plt.title('Distribution of Transportation Modes')
plt.xticks(rotation=0)
plt.show()
print("Business Insight: This chart shows which transportation modes are used most frequently in the supply chain.")
```

```python
shipping_cost = df.groupby('Transportation modes')['Shipping costs'].mean()

plt.figure(figsize=(8, 5))
shipping_cost.plot(kind='bar')
plt.xlabel('Transportation Mode')
plt.ylabel('Average Shipping Cost')
plt.title('Average Shipping Cost by Transportation Mode')
plt.xticks(rotation=0)
plt.show()
print("Business Insight: Comparing average shipping costs can help identify more cost-efficient transportation modes.")
```

```python
for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    print(column, ":", len(outliers), "outliers")
```

```python
correlation = df[numerical_columns].corr()

correlation
plt.figure(figsize=(12, 8))
plt.imshow(correlation, cmap='coolwarm', aspect='auto')
plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title('Correlation Matrix')
plt.show()
```

```python
print("BUSINESS INSIGHTS")
print("-" * 50)

print("1. Product types differ in their average revenue and sales volume.")
print("2. Customer demographics contribute differently to overall revenue.")
print("3. Transportation modes have different average shipping costs.")
print("4. Revenue, price, manufacturing cost, and sales volume show different levels of variation.")
print("5. Boxplots help identify potential outliers in revenue and manufacturing costs.")
print("6. The correlation analysis helps identify relationships between numerical supply-chain variables.")
```
